### Callable Class Attributes

Class attributes can be any object type, including callables such as functions:

```python
class Program:
    language = 'Python'
    
    # Bad approach - don't do this:
    # def say_hello():
    #     print(f'Hello from {Program.language}!')
    # Problems with this approach:
    # 1. Doesn't follow Python's instance method convention
    # 2. Will raise TypeError when called on instances
    # 3. Hard-codes class name, making inheritance difficult
    # 4. Can't access instance attributes if needed later
    
    # Good approach - use self parameter:
    def say_hello(self):
        print(f'Hello from {self.language}!')
```

```python
Program.__dict__
```

As we can see, the `say_hello` symbol is in the class dictionary.

We can also retrieve it using either `getattr` or dotted notation:

```python
Program.say_hello, getattr(Program, 'say_hello')
```

And of course we can call it, since it is a callable:

```python
# This will raise TypeError because say_hello() requires a 'self' parameter
# Even when called through the class, we need to provide an instance as the first argument
# Program.say_hello() tries to call it with no arguments
Program.say_hello

# These commented lines show other invalid ways to call it:
# print(say_hello())  # NameError - not in global namespace
# Program.say_hello   # Gets the function but doesn't call it
# Program.say_hello() # This will raise TypeError because say_hello() requires a 'self' parameter - we need to call it on an instance
```

```python
# create myProgram and call say_hello
# This doesn't work because say_hello() is defined without a 'self' parameter.
# When called on an instance (myProgram.say_hello()), Python automatically passes
# the instance as the first argument, but our method isn't set up to receive it.
# To fix this, we need to either:
# 1. Add 'self' parameter: def say_hello(self):
# 2. Use @staticmethod decorator
# 3. Call it through the class: Program.say_hello()
myProgram = Program()
# This raises TypeError: Program.say_hello() takes 0 positional arguments but 1 was given
myProgram.say_hello()  

```

We can even access it via the namespace dictionary as well:

```python
# access the say_hello method from the namespace dictionary
Program.__dict__['say_hello']()

# This will raise TypeError because say_hello() requires a 'self' parameter
# Even when called through the class, we need to provide an instance as the first argument
# Program.say_hello() tries to call it with no arguments
Program.say_hello
    
```
