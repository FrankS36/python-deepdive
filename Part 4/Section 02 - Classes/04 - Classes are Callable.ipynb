### Classes are Callable

As we saw earlier, one of the things Python does for us when we create a class is to make it callable.

Calling a class creates a new instance of the class - an object of that particular type.

```python
class Program:
    language = 'Python'
    
    def say_hello():
        print(f'Hello from {Program.language}!')
```

```python
p = Program() # this creates an instance of the Program class (instantiates the class)
```

## Namespace of Classes and Instances

When we create a class in Python, it has its own namespace that contains all the class attributes and methods.
This namespace is accessible via the class's `__dict__` attribute.

When we create an instance of a class, that instance gets its own separate namespace, also accessible via `__dict__`.
The instance namespace initially starts empty, and we can add instance attributes to it.

While the instance has its own namespace, it can still access the class attributes through the class namespace.
When we try to access an attribute on an instance, Python first looks in the instance namespace,
and if not found there, it then looks in the class namespace.

This creates a hierarchy of namespaces, where the instance namespace has priority over the class namespace.



```python
type(p) #__main__.Program means that p is an instance of the Program class
```

```python
isinstance(p, Program)
```

These instances have their own namespace, and their own `__dict__` that is distinct from the class `__dict__`:

```python
p.__dict__
```

```python
# The __dict__ of a class returns a mappingproxy object
# This is a read-only dictionary proxy that prevents direct modification
# of the class namespace dictionary for safety reasons
# It contains all the class attributes, methods and other special attributes
Program.__dict__
```

Instances also have attributes that may not be visible in their `__dict__` (they are being stored elsewhere, as we'll examine later):

```python
p.__class__
```

Although we can use `__class__` we can also use `type`:

```python
type(p) is p.__class__
```

Generally we use `type` instead of using `__class__` just like we usually use `len()` instead of accessing `__len__`.

Why? Well, one reason is that people can mess around with the `__class__` attribute:

```python
class MyClass:
    pass
```

```python
m = MyClass()
```

```python
type(m), m.__class__
```

But look at what happens here:

```python
class MyClass:
    __class__ = str
```

```python
m = MyClass()
```

```python
type(m), m.__class__ # __class__ is now a str, not MyClass
```

So as you can see, `type` wasn't fooled! Even though we redefined the `__class__` attribute to be the string class, `type()` still correctly identifies the true type of the object.

This is because `type()` looks at the actual type of the object in memory, while `__class__` is just an attribute that can be modified. This is one reason why it's generally safer to use `type()` rather than accessing `__class__` directly when you need to check an object's type.
