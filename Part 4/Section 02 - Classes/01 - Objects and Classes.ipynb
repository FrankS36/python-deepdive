### Objects and Classes

A class is a type of object. In Python we create classes using the `class` keyword.

```python
class Person:
    pass
```

```python
class Program:
    language = "Python"  # Class attributes instead of instance attributes
    version = "3.10"
    author = "Guido van Rossum"
    release_date = "2022-10-04"
    license = "PSF" 
    license_url = "https://www.python.org/psf/license/"
    license_text = "Python Software Foundation License"
    
    def __init__(self):
        pass  # No need to set instance attributes since we're using class attributes

# view the attributes of the class
print(Program.__dict__)

# view the name of the class
print(Program.__name__)

print(Program.__init__)

print(getattr(Program, '__init__'))

print(getattr(Program, 'language'))

print(getattr(Program, 'version'))

# Note on getattr() vs dot notation:
# 1. Dot notation (obj.attr) is more concise but requires attribute name to be known at write time
print(Program.language)  # Direct dot notation

# 2. getattr() is more flexible as it can use variables/strings for attribute names
attr_name = 'language'
print(getattr(Program, attr_name))  # Same result as Program.language

# 3. getattr() allows setting a default value if attribute doesn't exist
print(getattr(Program, 'nonexistent', 'default'))  # Returns 'default'

# 4. Dot notation raises AttributeError if attribute doesn't exist
try:
    print(Program.nonexistent)
except AttributeError as e:
    print(f"AttributeError caught: {e}")


```

Classes have "built-in" attributes, even though we did not specifically add any to the class ourselves.

For example, they have a name:

```python
Person.__name__
```

They are also callables, and calling a class results in the creation and return of a new **instance** of that class:

```python
p = Person()
```

Now the type of the object is the class used to build that object:

```python
type(p)
```

These instances also have "built_in" properties, which we will cover throughout this course.

For example, they have a `__class__` property that tells us which class was used to create the instance:

```python
p.__class__
```

As you can see that returns the class object used to instantiate `p`.

In fact:

```python
type(p) is p.__class__
```

We can also use `isinstance` to test if an object is an instance of a particular class - now this gets a bit more complicated when we use inheritance, but right now we're not, so it's quite straightforward:

```python
isinstance(p, Person)
```

```python
isinstance(p, str)
```

We can even use `isinstance` with our class, since we know it's type is `type`:

```python
isinstance(Person, type)
```

`type` is like the most generic kind of **class** object - we'll come back to this when discussing meta programming.

We really need inheritance to understand how this works, but every class **is** a `type` object (it inherits all the properties of `type`).

For now let's just see what functionality `type` has:

```python
help(type)
```

As you can see it has a `__call__` method (that's how our class becomes callable), and a bunch of other attributes and methods that we'll see throughout this course.

Our class objects also have these properties, because they inherit from the `type` object.

And in fact, `type` is an instance of itself - that's kind of weird, and not the case for our own classes:

```python
isinstance(type, type)
```

```python
# python is a dynamic language, we can add attributes to objects dynamically 
# dynamicallly means we can add attributes to objects at runtime
# Note: Although Python allows dynamically adding attributes, it's generally not 
# considered good practice because:
# 1. It makes code harder to understand and maintain
# 2. It bypasses the normal class definition structure
# 3. It can lead to bugs since attributes may or may not exist at runtime
# 4. Static type checkers and IDEs can't detect these dynamic attributes
# 5. It's better to define all attributes in the class definition or __init__


class Animal:
    def __init__(self, name, species):
        self.name = name
        self.species = species


setattr(Animal, 'diet', 'carnivore')

print(Animal.diet)

# exception to when we can dynamically add attributes to objects:
# 1. When working with data from external sources (e.g., JSON, databases) where
#    attributes are not known until runtime
animal_data = {'habitat': 'jungle', 'age': 5, 'weight': 100}
for key, value in animal_data.items():
    setattr(Animal, key, value)

# 2. When creating plugins or extension systems where new functionality
#    needs to be added dynamically

# 3. In testing/mocking scenarios where we need to temporarily add attributes
#    or methods for testing purposes

# 4. In metaprogramming use cases where class attributes are determined
#    programmatically based on some logic or configuration

# However, even in these cases, it's better to:
# - Document the dynamic nature of these attributes
# - Consider using __slots__ to restrict attribute addition
# - Use dataclasses or similar structured approaches when possible
# - Implement proper error handling for missing attributes
```

```python
# remove an attribute dynamically (at runtime)
delattr(Animal, 'diet')

# check if an attribute exists
hasattr(Animal, 'diet')


# There are several valid use cases for dynamically removing attributes:

# 1. Cleanup in testing scenarios
# When mocking objects for tests, we may add temporary attributes
# that need to be removed afterwards
class TestClass:
    test_attr = "temporary"
    
# Later in cleanup:
delattr(TestClass, 'test_attr')

# 2. Plugin/extension systems
# Disabling or unloading plugin functionality
class Plugin:
    feature_enabled = True
    
# When disabling plugin:
delattr(Plugin, 'feature_enabled')

# 3. Runtime optimization
# Removing unused attributes to free memory
class HeavyClass:
    large_data = [1, 2, 3] * 1000000
    
# When data is no longer needed:
delattr(HeavyClass, 'large_data')

# 4. Security considerations
# Removing sensitive data after use
class SecureClass:
    api_key = "secret123"
    
# After authentication:
delattr(SecureClass, 'api_key')

# However, like dynamic attribute addition, attribute removal should be:
# - Used sparingly and with good reason
# - Well documented
# - Handled safely with proper error checking
# - Considered carefully for impact on code maintainability


```

```python
# Let's explore how attributes are stored in a class

class Person:
    name = "John"
    age = 30
    
    def greet(self):
        return f"Hello, I'm {self.name}"

# The class's attributes are stored in its __dict__ attribute
print("Class __dict__:", Person.__dict__)

# Create an instance
p = Person()

# Instance attributes are stored in the instance's __dict__
print("\nEmpty instance __dict__:", p.__dict__)

# Add an instance attribute
p.height = 175

print("\nInstance __dict__ after adding height:", p.__dict__)

# Note: Not all attributes are stored in __dict__
# Some special attributes like __slots__ or attributes from C extensions
# may be stored differently. But for most Python-defined attributes,
# __dict__ is where they live.

# We can also use vars() to view the __dict__:
print("\nUsing vars() on instance:", vars(p))
print("Using vars() on class:", vars(Person))


```
